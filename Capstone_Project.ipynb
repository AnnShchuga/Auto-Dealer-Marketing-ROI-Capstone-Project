```python
import io
import pandas as pd
import numpy as np
import plotly.express as px

#df = pd.read_csv(io.BytesIO(uploaded['dataset.csv']))
df_users = pd.read_excel('/content/drive/MyDrive/ГикБреинс/Итоговая/Данные для Итоговой аттестации игровой аналитики.xlsx',
                   sheet_name='Информация о пользователях1')
df_company = pd.read_excel('/content/drive/MyDrive/ГикБреинс/Итоговая/Данные для Итоговой аттестации игровой аналитики.xlsx',
                   sheet_name='Таблица по компаниям')

display(df_users)
display(df_company)
```

```python
# Checking for duplicate rows
df_users[df_users.duplicated()].count()
```

```python
print('Total lines in df:', df_users.shape[0])
```

```python
# Clearing data, setting logical integrity and interconnection in data
df_clean = df_users.copy()

df_clean = df_clean.drop(['OS_Version' ,'Unnamed: 13', 'Unnamed: 14', 'Unnamed: 15', 'Unnamed: 16', 'Unnamed: 17', 'Unnamed: 18'], axis=1)
df_clean['Project_Name'] = df_clean['Project_Name'].str.replace(' ','', regex=True).str.replace('|','', regex=True)
df_clean['Country'] = df_clean['Country'].str.replace(' ','', regex=True).str.replace('|','', regex=True)
df_clean['Campaign'] = df_clean['Campaign'].str.replace(' ','', regex=True).str.replace('|','', regex=True)
df_clean['Organic'] = df_clean['Organic'].str.replace(' ','', regex=True).str.replace('Organic','Yes', regex=True)
df_clean.loc[~df_clean['Organic'].isin(['Yes', 'No']), 'Organic'] = np.nan

df_clean['OS'] = df_clean['OS'].str.replace(' ','', regex=True).str.replace('\d+', '', regex=True)
df_clean.loc[df_clean['OS'].isin(['OS']), 'OS'] = 'iOS'

projects = df_company['Project_Name'].drop_duplicates()
df_clean = df_clean[df_clean['Project_Name'].isin(projects)]

df_clean['Install_Date'] = pd.to_datetime(df_clean['Install_Date'], errors="coerce")
df_clean['Revenue (USD)'] = pd.to_numeric(df_clean['Revenue (USD)'], errors = "coerce")
df_clean['Time_In_Game (hrs)'] = pd.to_numeric(df_clean['Time_In_Game (hrs)'], errors = "coerce")
df_clean['Sessions'] = pd.to_numeric(df_clean['Sessions'], errors = "coerce")

df_clean = df_clean.dropna()

print('Total rows in the cleared data:', df_clean.shape[0])
print('The data loss is:', (1 - df_clean.shape[0] / df_users.shape[0]) * 100, '%')
```

We get a dataset without empty, duplicate, or incorrect data.

```python
#Calculating the Installs metric as a whole and for each project

df = df_clean.copy()

df_installs = df.groupby(['Install_Date'], as_index=False).agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'Installs'})

df_installs_game1 = df[(df['Install_Date']>='2023-07-01') & (df['Project_Name'] == 'Pirates_Adventure')].groupby(['Install_Date'], as_index=False) \
                    .agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'Installs'})
df_installs_game2 = df[(df['Install_Date']>='2023-07-01') & (df['Project_Name'] == 'Forbidden_Forest')].groupby(['Install_Date'], as_index=False) \
                    .agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'Installs'})
df_installs_game3 = df[(df['Install_Date']>='2023-07-01') & (df['Project_Name'] == 'Mythic_Dragons')].groupby(['Install_Date'], as_index=False) \
                    .agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'Installs'})
df_installs_game4 = df[(df['Install_Date']>='2023-07-01') & (df['Project_Name'] == 'Galactic_Wars')].groupby(['Install_Date'], as_index=False) \
                    .agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'Installs'})
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
fig = make_subplots(rows=2, cols=1, subplot_titles=("All time", "Older 1 Jun 2023"))

fig.add_trace(go.Scatter(x=df_installs['Install_Date'], y=df_installs['Installs'], name='All time', legendgroup=1, legendgrouptitle_text="All time"),row=1, col=1)

fig.add_trace(go.Bar(x=df_installs_game1['Install_Date'], y=df_installs_game1['Installs'], name='Pirates_Adventure', legendgroup=2,
                     legendgrouptitle_text="Older 1 Jun 2023"),row=2, col=1)
fig.add_trace(go.Bar(x=df_installs_game2['Install_Date'], y=df_installs_game2['Installs'], name='Forbidden_Forest', legendgroup=2),row=2, col=1)
fig.add_trace(go.Bar(x=df_installs_game3['Install_Date'], y=df_installs_game3['Installs'], name='Mythic_Dragons', legendgroup=2),row=2, col=1)
fig.add_trace(go.Bar(x=df_installs_game4['Install_Date'], y=df_installs_game4['Installs'], name='Galactic_Wars', legendgroup=2),row=2, col=1)

fig['layout'].update(height=700, title="Installs")
fig.show()
```

The surge in installations occurred after July 1, 2023 for all projects. Perhaps before that, the games were at the testing stage and were not widely available.

Information on installations is provided in 5-day increments. Perhaps the statistics module provides data with such frequency.

The DAU metric will match the Installs metric, for the same reason that we cannot calculate Retention - there is no event date to count against. The same applies to the PPU metric.

The DAU and PPU metrics will be calculated for all projects in general.

```python
# DAU (Daily Active Users)
df_dau = df.groupby('Install_Date', as_index=False).agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'dau'})

# PPU (Percentage of Paying Users (the share of the paying audience relative to DAU))
df_payers_day = df[df['Revenue (USD)']>0].groupby('Install_Date', as_index=False).agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'payers'})
df_ppu = pd.merge(df_dau, df_payers_day, on='Install_Date', how='left')
df_ppu['payers'] = df_ppu['payers'].fillna(0)
df_ppu['ppu'] = round(df_ppu['payers'] / df_ppu['dau'], 2)
df_ppu
```

```python
fig_ppu = px.line(df_ppu, x='Install_Date', y='ppu')
fig_ppu.show()
```

According to the received PPU, it can be concluded that the motivations for payment by users were chosen correctly.

```python
#Let's analyze the installations relative to the organic and geographical cohorts, also limiting ourselves to the data obtained after the peak of installations on 07/01/2023.

df_installs_organic = df[df['Install_Date']>='2023-07-01'].groupby(['Install_Date', 'Organic'], as_index=False).agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'Installs'})

fig = px.line(df_installs_organic, x = df_installs_organic['Install_Date'], y = df_installs_organic['Installs'], color=df_installs_organic['Organic'])
fig.show()
```

From the graph, we can conclude that the unpaid and paid trafixes are approximately equal in terms of the number of installations.

```python
#Let's look at the geography of the installations

df_installs_world = df[df['Install_Date']>='2023-07-01'].groupby(['Country'], as_index=False) \
                    .agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'Installs'}).sort_values(by='Installs', ascending=False)
df_installs_world.head(10)
```

```python
#Let's look at these data regarding the paid and free traffic of the top 20 countries by installations.

df_installs_world_organic = df[(df['Install_Date']>='2023-07-01') & (df['Country'].isin(df_installs_world['Country'].head(20).values))] \
                    .groupby(['Country', 'Organic'], as_index=False) \
                    .agg({'User_ID':'nunique'}).rename(columns = {'User_ID':'Installs'})

fig = px.bar(df_installs_world_organic, x = 'Country', y = 'Installs', color='Organic', barmode = 'group')
fig.show()
```

According to the graph, the shares of paid and free traffic are about the same. Therefore, it can be concluded that the advertising campaigns were able to attract about the same number of users as they came on their own.

```python
#Counting the CPI

df_spend = pd.merge(df[df['Install_Date']>='2023-07-01'], df_company, how='left', on=['Campaign', 'Project_Name'])

df_spend['user_cost'] = np.where(df_spend['Organic'] == 'Yes', df_spend['Advertising_Costs (USD)'] / df_spend['Clicks'], 0)
df_spend = df_spend[df_spend['Install_Date']>='2023-07-01'].groupby(['Project_Name', 'Campaign'], as_index=False) \
           .agg({'User_ID':'nunique', 'Revenue (USD)': 'sum', 'user_cost': 'sum'}) \
           .rename(columns = {'User_ID':'Installs', 'Revenue (USD)': 'revenue', 'user_cost':'spend_daily'})
df_spend['cpi'] = df_spend['spend_daily'] / df_spend['Installs']
df_spend[df_spend['cpi'] > 0]
```

Of all the various advertising campaigns, we have cost data for only 8 campaigns. Based on the CPI, we can say that the campaigns turned out to be quite successful.

```python
#Consider the effectiveness of advertising campaigns for projects
#Consider revenue from advertising campaigns over $1,000 as the most effective

fig = px.bar(df_spend[df_spend['revenue']>=1000], x = 'Campaign', y = 'revenue', color='Project_Name', barmode = 'group')
fig['layout'].update( title="Revenue")
fig.show()
```

```python
#Counting LTV and ROI

df_ltv = pd.merge(df[df['Install_Date']>='2023-07-01'], df_company, how='left', on=['Campaign', 'Project_Name'])

df_ltv['user_cost'] = np.where(df_ltv['Organic'] == 'Yes', df_ltv['Advertising_Costs (USD)'] / df_ltv['Clicks'], 0)
df_ltv = df_ltv.groupby(['Install_Date'], as_index=False).agg({'User_ID':'nunique', 'Revenue (USD)': 'sum', 'user_cost': 'sum'}) \
           .rename(columns = {'User_ID':'Installs', 'Revenue (USD)': 'revenue', 'user_cost':'spend_daily'})

df_ltv['install_month'] = df_ltv['Install_Date'].dt.month
df_ltv = df_ltv.groupby(['install_month'], as_index = False).agg({'Installs':'sum', 'spend_daily':'sum', 'revenue':'sum'}) \
         .rename(columns = {'spend_daily':'spend_month', 'revenue': 'total_revenue'})
df_ltv['ltv'] = df_ltv['total_revenue'] / df_ltv['Installs']
df_ltv['cpi'] = df_ltv['spend_month'] / df_ltv['Installs']
df_ltv['roi'] = df_ltv['total_revenue']*100 / df_ltv['spend_month']

df_ltv
```

We can conclude from the LTV of all projects that almost all users generated revenue, respectively, LTV turned out to be quite good.

According to the ROI metric, it can be said that the payback of advertising campaigns turned out to be high.
